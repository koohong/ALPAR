# ALPAR - Automated Learning Pipeline for Antimicrobial Resistance

Automated Learning Pipeline for Antimicrobial Resistance

![Pipeline](docs/imgs/ALPAR_Flowchart_new.png)

## Description 

- `Antimicrobial resistance (AMR)` happens when bacteria evolve so that antibiotics no longer work against them, making infections harder to treat.

- This paper introduces ALPAR, a computer tool that combines genetics, bioinformatics, and machine learning to predict whether bacteria will be resistant to antibiotics.

   - Normally, studying antibiotic resistance requires many different software programs and technical steps, which can be difficult and time-consuming. ALPAR automates most of the process in a single pipeline.

- Researchers can give ALPAR bacterial genome sequences, and it automatically finds genetic mutations, builds data tables, trains machine learning models, and identifies genes linked to resistance.

- The system also includes safeguards to reduce common mistakes, such as models learning from closely related bacteria instead of finding the true causes of resistance.

- To test ALPAR, the researchers analyzed about 1,000 E. coli genomes and used machine learning to predict resistance to the antibiotic ciprofloxacin.

    - The tool successfully identified well-known resistance mutations in genes such as gyrA and parC, showing that it was finding biologically meaningful patterns.

    - ALPAR performed much better than a traditional `genome-wide association study (GWAS)` approach on this dataset.

    - The tool was also tested in international antimicrobial resistance prediction competitions (CAMDA), where `it won the 2024 challenge and` placed `third in the 2025 challenge`.

- When compared with an established rule-based resistance prediction tool, ALPAR achieved better overall prediction performance across many bacterial species and antibiotics.

- The authors conclude that ALPAR can help scientists more quickly discover resistance-related mutations and make accurate predictions from bacterial DNA data.

   In simple terms, ALPAR is like a smart assistant that reads bacterial DNA and helps researchers predict which antibiotics are likely to fail, potentially supporting better treatment decisions and the fight against antibiotic-resistant infections.

    - The installable CLI is `alpar`. The Python package on disk is still named `sr_amr` (from the earlier "Single-Reference AMR" name).

## Windows: Linux/macOS-only tools require WSL

- **You need WSL to run this pipeline on Windows.** 
  - Native Windows can browse the code, read `tool_results/`, and 
  - maybe install the Python-only deps via pixi. 
  - It cannot run variant calling, annotation, trees, or the full pipeline.

![Starting WSL from PowerShell](fig/1.png)

- `automatix` / `create_binary_tables` create extra conda environments on demand for tools such as `snippy`, `prokka`, `cd-hit`, `panaroo`, `mashtree`, `bakta`, `pyseer`, and `panacota` (see [sr_amr/envs](sr_amr/envs)). 

- Those packages and their dependencies (e.g. `bcftools`, `aragorn`) are **only published for Linux/macOS on bioconda and have no `win-64` build**. Environment creation on native Windows fails with errors like `nothing provides bcftools` or `PackagesNotFoundError`.

To run ALPAR on a Windows machine, use **WSL2 (Windows Subsystem for Linux)**.

**Where to run each command:**

  - The Pixi installer (`curl ... | sh`), `pixi install`, and
  - `alpar` must run in an **Ubuntu** terminal, not PowerShell and not a Cursor PowerShell terminal. That `curl` script is a Linux installer.

## Windows copy vs Ubuntu copy

- A clone or copy in Ubuntu (`~/ALPAR`) is a **second, separate folder** from the ALPAR that Cursor has open on Windows (for example `C:\Users\...\Documents\GitHub\ALPAR`). 
- They are not the same files.

| | Windows (Cursor / OneDrive) | Ubuntu |
|---|---|---|
| Path | `C:\Users\...\GitHub\ALPAR` | `/home/<user>/ALPAR` (`~/ALPAR`) |
| Seen in Ubuntu as | `/mnt/c/Users/.../GitHub/ALPAR` | `~/ALPAR` |
| Same files? | No — two copies | No |

Edits in one do **not** appear in the other unless you `git push` / `git pull` or copy files yourself.

If you already have the Windows repo, you do not have to clone again. Copying still creates a second copy:

```shell
mkdir -p ~/ALPAR
cp -a "/mnt/c/Users/<you>/OneDrive - Danaher/Documents/GitHub/ALPAR/." ~/ALPAR/
```

Pick one folder to edit day to day so the two copies do not drift.

- Pushing the Ubuntu copy to GitHub
    - The Ubuntu copy can push to GitHub the same way the Windows copy can, as long as it is a git repo with a remote.

- **`git clone` into `~/ALPAR`:** `origin` is already set. After commits: `git push -u origin HEAD`
- **`cp -a` from OneDrive including `.git`:** remotes and history come along. You can push from either folder.
- **No `.git` folder:** it is just files. Clone instead, or `git init` and `git remote add origin <url>` — that is a new repo, not a continuation of the Windows one.

You need GitHub login **inside Ubuntu** (HTTPS prompt, `gh auth login`, or an SSH key in WSL). Windows credentials do not always carry over.

Do not push both copies to the same branch with different un-pulled commits. Pick one folder to commit from, or always `git pull` before `git push`.

# Sample Data Selection 

Local tables in `data/cabbage/` from the [EMBL-EBI AMR Portal](https://www.ebi.ac.uk/amr/) ([CABBAGE](https://doi.org/10.1093/nar/gkag780)).

| File | Grain |
|---|---|
| `phenotype.parquet` | one isolate × one antibiotic AST result |
| `genotype.parquet` | one isolate × one predicted AMR gene/element |
| `phenotype_genotype_merged.parquet` | inner join of the two on `BioSample_ID`, `assembly_ID`, `antibiotic_ontology` |

Folder layout and samples: [0_description_of_data.md](../0_description_of_data.md).

The two parent tables are large (~1.7M and ~1.4M rows). This notebook reads **only the columns it needs** and aggregates before plotting.

- `Phenotype` is a lab result.  Someone grew the isolate and ran an antimicrobial susceptibility (AST) to call it susceptible, intermediate or resistant.
- `Genotype` is what is encoded in the DNA.  You might find an AMR gene or mutation associated with AMR.
   -  genotype $\not =$ phenotype.  Having a resistance gene doesn't necessarily guarantee that the organisim will show resistance.
   -  Gene expression, mutation, regulartory mechanisms and other biological factors can affect the observed phenotype.
   -  `Genome sequence` > `genetic features` > `ML model` > `predicted phenotype`.
   -   Each row in genotype.parquet is one known antimicrobial-resistance gene or element that `AMRFinderPlus` found in that isolate’s genome assembly.


# Iminpenem example



The files under Resistant and Susceptible are the strains being tested. reference.gbff is the single baseline genome they are all compared with. The folder names are phenotype labels, not a description of how those files differ from the reference.

|     |Resistant/*.fna and Susceptible/*.fna	|reference.gbff|
|:=====|:---------::-------------------------|
|What it is | One assembled genome per strain, FASTA | One annotated reference genome, GenBank|
|How many   |Many,  one file per strain     | One file for the whole run|
|What the name means| Measured result for that antibiotic: resistant is coded 1, susceptible is coded 0| No phenotype| 

- ALPAR does not treat it as resistant or susceptible
     - `Snippy` aligns every .fna to that same reference (snippy --ctgs --reference reference.gbff <strain.fna>) and records where the strain differs from it.
     - Those differences become the mutation columns.
     - `Prokka` also uses the reference, as the protein annotation source (--proteins reference.gbff).

A mutation column therefore means "this strain differs from the reference here." Both resistant and susceptible strains are measured against that same baseline. The model then looks for which of those differences line up with the 1 / 0 labels from the folder names.


## Prep Files

### .fna and .gbff

- One isolate is one `.fna`.  This file contains many contigs and ALPAR calls this file one strain.
- Isolate is the biological sample: one bacterial culture.
  - In the `CABBAGE` tables that is a BioSample such as SAMN04550082.
  - In this ALPAR run the isolate is whichever CAMDA genome was copied in, named like Aba_IMI_R_128.

- `.fna` is that isolate’s assembled genome in FASTA format.
    - One file goes in as one input to Snippy, and it becomes one row in the binary mutation table and the phenotype table.

- `Contigs` are the separate DNA pieces inside that file.
    - Each piece is one FASTA record, a `>` header followed by sequence.
    - A finished genome can be a single contig.
    - These CAMDA assemblies are drafts, so one `.fna` usually contains many contigs that the assembler could not join.
    - Contigs assembly only joins reads into DNA pieces.  It does not name genes.

- ALPAR always runs `Snippy` with `--ctgs`, which tells Snippy the file is an assembly of contigs rather than raw reads.

- `Strain`, in these ALPAR tables, is the ID for that one .fna.
  - It is the short name from random_names.txt (Aba_IMI_R_128 → DKDR9B6J). It is not a separate biological object from the isolate.

- So: one isolate → one .fna → many contigs inside it → one Strain row.

- Each input `.fna` is one isolate’s `de novo` assembly: the contigs written after the reads were assembled.
- The extension itself only means nucleotide FASTA.
- A reference chromosome can be saved as .fna too.
- Here the reference is the `.gbff`, and the `.fna` files are the assembled isolate genomes Snippy compares to it.

### random_names.txt.  

ALPAR always renames every genome to a short ID before it runs the tools. random_names.txt is the lookup that keeps the original .fna name.

Each name is 8 characters, letters and digits only, like DKDR9B6J. That string becomes the Snippy output folder, the Prokka or Bakta output folder, the Bakta filename prefix, and the Strain column in binary_mutation_table.tsv. Prokka is started with --compliant, which follows GenBank naming rules, and Bakta uses the ID as --prefix. A short alphanumeric ID stays valid for those tools when the original filename is long or contains characters those programs reject

### phenotype_table.tsv

- ALPAR built that phenotype table from the input folders. A file under imipenem/Resistant/ is 1. A file under imipenem/Susceptible/ is 0. It then replaced the filename with the short ID from random_names.txt. All 87 rows match.

# Bioinformatics software programs and tools

## AMRFinderPlus

- This not part of Snippy or Prokka run.
- AMRFinderPlus searches each assembled genome against a catalog of known resistance genes and reports a hit when the sequence matches. Those hits are the rows in genotype.parquet: one known AMR gene per row, with names such as tetA or blaEC. It only reports genes already in that catalog.

Snippy and Prokka do not do that screen. Snippy records any difference from the reference. Prokka names whatever coding sequences it finds. A resistance gene can show up in those results, but only because it is a variant or a protein, not because a tool checked it against an AMR catalog.

CARD is the other known-AMR catalog in this repo. ALPAR does not run it.

- AMRFinderPlus searches the assembled contigs directly for known resistance genes.
    - Those results are already stored in genotype.parquet, from the CABBAGE pipeline.
    - It takes de novo assembly as input.  

## assembler

- From FASTQ file, an `assembler` creates FASTA.
- The step is called `de novo assembly`, and the program is an `assembler`.
    - There is no single assembler that every FASTQ file goes through.
- A `FASTQ` file holds the `raw reads` and a `quality score` for each base.
    - The assembler overlaps those reads and writes the joined pieces as `FASTA`.
    - Those pieces are the `contigs`.
    - A nucleotide `FASTA` is often named `.fna`, which is the file ALPAR uses.

- ALPAR never sees the `FASTQ`.

- For bacterial genomes like these, a common assembler is `SPAdes`.
    - Others include `SKESA`, `Unicycler`, and `MEGAHIT`. Which one was used is decided by whoever built that assembly, and this repo does not record it.

- De novo assembly builds the genome from the reads alone. It does not look at a reference.
- `Snippy` takes that finished assembly, the `.fna of contigs`, and aligns it to the reference `.gbff`.
    - Differences found in that alignment are the variants in snps.vcf.

## Snippy

|Input | Output | description |
|:-----|:-------|:---------|
|.fna, .gbff | snps.tab| SNP location and description |
|snps.tab    | snps.vcf| check if the SNP falls within gene.  Only the variant itself is saved in snps.vcf.  The gene, effect, and product are not|
|snps.tab    |mutations_annotations.tsv | ALPAR reads snps.vcf to mark each variant present or absent and snps.tlab to attach the gene.  This is a catelog of unique variants, not a per-genome list.  One row per unique variant, with gene effect, and product|
|snps.vcf   | binary_mutation_table.tsv | this output stores

- Snippy lines the sample .fna up against the single .gbff. (Snippy aligns each de novo assembly to the reference)
- Whereever the bases do not match, it records a varianit (i.e., place where this isotlate's DNA differs from the reference genome), the reference coordinate, the reference base, and the base in this sample.
  - A typical SNP looks like `position 1,234,456 reference A, this isolate G`.
  - Insertions and deletions are variants too.  `snps.fcf` is that list

- Snippy compares one assembly at a time to the reference. It records each place that assembly differs. If that position sits inside a gene already marked on the reference, it attaches that gene name. A mutation found in only one sample is still recorded. ALPAR later stacks those per-sample lists into mutations_annotations.tsv and the binary table.

### mutations_annotations.tsv

- Snippy then asks `does position 1,234,456 fall inside a gene?` and `snps.tab` keeps the same variance and adds that lookup.
  - Snippy does not keep snps.tab and snps.vcf.  Only `mutations_annotations.tsv` is saved
  -  A gene that sits entirely inside one contig is called there. If a contig break falls in the middle of a gene, the caller usually sees two partial genes or misses it.
  -  Snippy gene names in mutations_annotations.tsv come from the reference `.gbff`.
    - Those genes were annotated on the reference after that reference genome was assembled.
    - Snippy only asks which reference gene contains the variant coordidnate.
    - It does not find genes in the sample `.fna`
 
- Which .fna has it is in binary_mutation_table.tsv:
    -  Each row is one genome.  The row name is the `.fna` filename without the extention.
    -  In this run those names are BioSample IDs such as SAMN04550082, matching phenotype_table.tsv.
    -  Each column is one mutation, using the same text as the Mutation column in mutations_annotations.tsv.
    -  1 means that genome has the variant. 0 means it does not.

So '313','T:C','snp' belongs to every .fna whose row has a 1 in that column. Join the two tables on that mutation string when you want the gene and effect next to the genome.  

### binary_mutation_table

- binary_mutation_table.tsv is a presence/absence grid built only from each genome’s snps.vcf. Gene names are not in it.

`create_binary_tables` does this:

- Each .fna gets a short ID, saved in random_names.txt.
    - For example `Aba_IMI_R_128` becomes `DKDR9B6J`.
    - That short ID is the Strain value in the table.

- Snippy runs once per .fna against the reference genome and writes `snps.vcf` into a folder named with that short ID.
    - ALPAR reads every `snps.vcf` and creates `binary_mutation_table.tsv`
    - Each variant becomes a column named 'position','reference:alternate','type', the same string as the Mutation column in mutations_annotations.tsv.
    - It takes the union of those variants across all genomes, sorts them, and writes one column per variant.
    - Each genome that finished Snippy becomes one row.
      - A cell is 1 if that variant is in that genome’s VCF, and 0 if it is not.
      - Genomes whose Snippy run failed are left out.

- This file is variants only.
    - Gene presence and absence are added later, in binary_mutation_table_with_gene_presence_absence.tsv.

- To get back to the .fna, look up the Strain ID in alpar_out/random_names.txt. The first column is the filename without .fna; the second column is the ID used in the binary table.

### binary_mutation_table_with_gene_presence_absence.tsv

- ALPAR creates `binary_mutation_table_with_gene_presence_absence.tsv` by copying binary_mutation_table.tsv and appending gene columns.
- This run used the defaults, `Prokka` then `CD-HIT`, because run_alpar_steps.sh does not pass --only_variants or another gene tool.

- The steps are:
    - Prokka annotates each assembled .fna and writes that genome’s proteins.
    - CD-HIT clusters those proteins at 95% identity. Proteins in one cluster are treated as the same gene.
    - ALPAR makes a presence/absence matrix from the clusters. The column name is the representative protein. A strain is 1 if it has a protein in that cluster and 0 if it does not.
    - Those gene columns are appended to the right of the existing mutation columns. The strain rows stay the same.

The mutation columns are still the Snippy variants. The new columns are whether each gene is present in that genome.

- This table has 87 data rows and 231,519 columns.
    - Every cell after the first column is 1 or 0. The columns come from two tools and sit in two blocks.
    - `Column 1`, Strain. This is the short ID ALPAR assigned in random_names.txt, such as DKDR9B6J.
        - It is the same ID as in binary_mutation_table.tsv and alpar_out/phenotype_table.tsv.
    - The next 229,014 columns are Snippy variants.
        - ALPAR read every snps.vcf and made one column per variant.
        - The header is the position on the reference, the base change, and the type:
        - '1000014','G:A','snp' means reference position 1,000,014 changed from G to A, and the change is a SNP.
        - A cell is `1` when that variant is in that genome’s VCF and `0` when it is not.
        - These columns do not contain the gene name.
        - The `gene`, `effect`, and `product for the same variant string` are in mutations_annotations.tsv, which ALPAR built from snps.tab.
    - The last 2,504 columns are gene presence or absence.
        - `Prokka` annotated each `.fna`, wrote a protein for each coding sequence, and
        - `CD-HIT` grouped proteins that are at least 95% identical.
        - One cluster becomes one column.
        - The header is the representative protein’s name, taken from Prokka’s protein FASTA.
        - Because Prokka was given the reference proteins, those names look like:
            - proteins_YP_001083085_1_______DNA_replication_protein
        - A cell is `1` when that genome has a protein in the cluster and 0 when it does not.
        - This is not the Prokka locus tag (LDEGODNI_00001). Those locus tags are in cd-hit/protein_positions.csv.

## Prokka and Bakta 

- Genes in the sample are found when Prokka or Bakta annotates each assembled .fna.
- That scan is what feeds the gene presence/absence columsn.
- AMRFinderPlus, the tool behind `genotype.parquet` works the same way.  It searches the finihsed assembly for known resistance genes.
- Prokka runs separately on each isolate .fna. Its gene calls are the ones CD-HIT clusters for the gene presence/absence columns.
    - Those names are not what Snippy writes for a mutation.
- Prokka runs after Snippy for the same genome. Different genomes can overlap.
    - For one `.fna`, `run_variant_calling_and_annotation` calls `Snippy` first and, when that finishes, calls `Prokka` on the same file.
    - `Prokka` does not wait for every genome to finish `Snippy`.
    - A pool of workers handles many genomes at once, so one worker can be in `Prokka` while another is still in `Snippy`.

- Prokka and Bakta do the same job.
    - Each one reads an assembled genome and writes down the genes it contains.
    - ALPAR runs one of them on every .fna. The default is Prokka.

- For each contig, the tool finds coding sequences, translates them to proteins, and assigns a locus tag and a product name.
  - Prokka in this run also uses the reference proteins, so a match can be named with a reference id such as YP_001084105.
  - The output per genome is a GenBank file, a protein FASTA, and a feature table.

- ALPAR then clusters those proteins with CD-HIT.
    - A protein found in a genome becomes a 1 in `binary_mutation_table_with_gene_presence_absence.tsv`.
    - That is the gene presence/absence half of the table.

- `Bakta` is the alternate annotation tool.
    - It uses its own database and often assigns more specific product names.
    - You select it with `--annotation_tool bakta`.
    - Snippy’s gene names in `mutations_annotations.tsv` still come from the reference, not from Prokka or Bakta.

- Prokka finds conding sequence (i.e., gene that produces protein) and reports the protein product from that coding sequence.
- Prokka scans each assembly for coding sequences, whether or not they contain a mutation. For each coding sequence it writes the protein sequence and a product name, such as transcriptional regulator. Those proteins are what become the presence/absence columns.

### cd-hit

- `CD-HIT` groups nearly identical proteins so the same gene is not counted once per genome.

- `Prokka` writes a protein sequence for every coding sequence in every `.fna`.
    - The same gene in `87 genomes` would otherwise look like 87 different proteins.
    - `CD-HIT` compares those sequences and puts proteins that are at least 95% identical into one cluster. Each cluster is treated as one gene.
    - Prokka found about 2,514 coding sequences in each of the 87 assemblies, 218,718 proteins in total.
    - CD-HIT grouped sequences that are at least 95% identical, so the same gene from many genomes becomes one column.
    - The column is named after a single representative sequence, not after every genome’s own locus tag.

- Those representatives are YP_ ids because Prokka was given the reference proteins.
    - A coding sequence that matched a reference protein inherited that protein’s id and product, for example `YP_001083085.1` and `DNA replication protein`.
    - The per-genome locus tags, such as LDEGODNI_00001, were written to cd-hit/protein_positions.csv and were not used as column names.

- In `binary_mutation_table_with_gene_presence_absence.tsv` the gene columns are all 0.
    - ALPAR decides which genome has the gene by reading the text before the first _ in the protein name.
    - Every name starts with proteins, not with a strain id such as DKDR9B6J, so no genome was marked present. The Snippy mutation columns are unaffected.

- ALPAR turns each cluster into one column in binary_mutation_table_with_gene_presence_absence.tsv. The column name is the representative protein, the sequence CD-HIT picked to stand for that cluster. A genome is 1 if it contributed a protein to the cluster and 0 if it did not.
    - Prokka’s gene names are not in `mutations_annotations.tsv`.
    - That `GENE` column comes from the reference through Snippy.
    - Prokka’s names survive in `alpar_out/cd-hit/protein_positions.csv`.

- `Prokka` wrote a `GenBank` file per genome, with a locus tag and a product.
    - ALPAR turned those into one row per protein. The original per-genome `Prokka folder`, `alpar_out/prokka/`, was deleted at the end of the run.

- A row such as DKDR9B6J_LDEGODNI_00001_transcriptional_regulator is strain DKDR9B6J, Prokka locus tag LDEGODNI_00001, and product transcriptional regulator.

- The extra columns in `binary_mutation_table_with_gene_presence_absence.tsv` are a different label.
    - They are the `CD-HIT` cluster names taken from Prokka’s protein FASTA headers,
        - For example `proteins_YP_001084105_1_..._hemolysin_type_calcium_binding_protein`.
        - The `YP_ id` and the product text come from the reference proteins passed to Prokka, not from the LDEGODNI_00001 locus tag.
 

# Questions

## Why are gene names obtained using both Snippy and Prokka?

Snippy and Prokka answer two different questions, so ALPAR uses information from both.

- `Snippy` identifies which reference gene contains a detected mutation. Because each genome is compared with the same reference genome, a mutation at a particular position within a gene, such as dnaA, can be compared consistently across isolates. This allows ALPAR to capture resistance mechanisms associated with sequence variants or point mutations, such as mutations within gyrA.

- `Prokka`, on the other hand, annotates genes present in each isolate's assembled genome. CD-HIT can then cluster similar proteins across isolates to generate gene presence/absence features. This allows ALPAR to capture resistance mechanisms associated with acquired genes that may not exist in the reference genome, such as an acquired beta-lactamase gene. Such genes would not necessarily be represented by Snippy because Snippy focuses on variants relative to the reference genome.

Therefore, the two approaches provide complementary features:

- Snippy: mutation/variant information relative to a common reference
- Prokka + CD-HIT: gene presence/absence information from each isolate's assembly

The mutation table contains the Snippy-derived features, while the additional columns in binary_mutation_table_with_gene_presence_absence.tsv contain the Prokka/CD-HIT-derived gene presence/absence features.

- `Prokka` finds those genes contained in the isolate (stored as one .fna) on the assembly and writes one protein (i.e., the molecule made by translating a protein-coding gene) sequence for each coding gene. That is why one genome produces many proteins.

- `CD-HIT` then compares proteins across genomes.
    - Proteins with nearly the same sequence are treated as the same gene, and that group becomes one column in binary_mutation_table_with_gene_presence_absence.tsv.
    - A 1 means this genome has that protein, so it has that gene. A 0 means it does not.

- Genes that do not encode a protein, such as rRNA, are not in those columns. This table is built only from Prokka’s protein sequences.

# Automatic Pipeline

From genomic files, creates binary mutation and phenotype tables, applies thresholds, creates phylogenetic tree, conducts GWAS analysis, calculates PRPS score and trains machine learning models with conducting feature importance analysis and splitting data aginst information leakage with [DataSAIL](https://github.com/kalininalab/DataSAIL) against all the given antibiotics.

- Input, `-i`: Path of folder that have structure: input_folder -> antibiotic -> [Resistant, Susceptible]

    `````shell
    📦input_folder
    ┣ 📂antibiotic1
    ┃ ┣ 📂Resistant
    ┃ ┃ ┣ 📜fasta1.fna
    ┃ ┃ ┗ 📜fasta2.fna
    ┃ ┃ ┗ ...
    ┃ ┗ 📂Susceptible
    ┃ ┃ ┣ 📜fasta3.fna
    ┃ ┃ ┗ 📜fasta4.fna
    ┃ ┃ ┗ ...
    ┗ 📂antibiotic2
    ┃ ┣ 📂Resistant
    ┃ ┃ ┣ 📜fasta2.fna
    ┃ ┃ ┗ 📜fasta5.fna
    ┃ ┃ ┗ ...
    ┃ ┗ 📂Susceptible
    ┃ ┃ ┣ 📜fasta2.fna
    ┃ ┃ ┗ 📜fasta3.fna
    ┃ ┃ ┗ ...
    ┗ 📂...
    `````

- Output, `-o`: Output folder path, where the output will be stored. If path exist, `--overwrite` option can be used to overwrite existing output.

- Reference, `--reference`: Reference file path, accepted file formats are: `.gbk .gbff`

- Custom database (Optional), `--custom_database`: Fasta file path for protein database creation, can be downloaded from [UniProt](https://www.uniprot.org/) accepted file formats are: `.fasta`

Basic usage:
`````shell
alpar automatix -i example/example_files/ -o example/example_output/ --reference example/reference.gbff
`````

## Binary Table Threshold

Applies threshold to binary mutation table, and drops columns that has less than threshold percentage, useful to reduce sequencing errors in the data.

- Input, `-i`: Binary mutation table path

- Output, `-o`: Output folder path, where the output will be stored. If path exist, `--overwrite` option can be used to overwrite existing output.

- Threshold percentage, `--threshold_percentage`: Threshold percentage value to be used to drop columns. If column sum is less than this value, columns will be deleted from table

Basic usage:
`````shell
alpar binary_tables_threshold -i example/example_output/binary_mutation_table.tsv -o example/example_output/ 
`````

## GWAS

Runs GWAS analysis to detect important mutations in the data.

- Input, `-i`:  Binary mutation table path that is created via create_binary_tables command, can be found in create_binary_tables output path as `binary_mutation_table_with_gene_presence_absence.tsv` or `binary_mutation_table.tsv` or if threshold applied, can be found in binary_table_threshold output path as `binary_mutation_table_threshold_*_percent.tsv`

- Phenotype, `-p`:  Binary phenotype table path,  can be found in create_binary_tables output path as `phenotype_table.tsv` if `--create_phenotype_from_folder` is used. Can also created manually and used.

- Tree, `-t` : Phylogenetic tree path, can be found in panacota output path as `phylogenetic_tree.newick` or phylogeny output path as `phylogenetic_tree.tree`

- Output, `-o`: Output folder path, where the output will be stored. If path exist, `--overwrite` option can be used to overwrite existing output.

Basic usage:
`````shell
alpar gwas -i example/example_output/binary_mutation_table_with_gene_presence_absence.tsv -p example/example_output/phenotype_table.tsv -t example/example_output/phylogeny/phylogenetic_tree.tree -o example_output/
`````

## PRPS

Runs PRPS (Phylogeny-Related Parallelism Score) to detect the mutations are more likely associated with phylogeny rather than antimicrobial resistance.

- Input, `-i`:  Binary mutation table path that is created via create_binary_tables command, can be found in create_binary_tables output path as `binary_mutation_table.tsv` or if threshold applied, can be found in binary_table_threshold output path as `binary_mutation_table_threshold_*_percent.tsv`

- Tree, `-t` : Phylogenetic tree path, can be found in panacota output path as `phylogenetic_tree.newick` or phylogeny output path as `phylogenetic_tree.tree`

- Output, `-o`: Output folder path, where the output will be stored. If path exist, `--overwrite` option can be used to overwrite existing output.

Basic usage:
`````shell
alpar prps -i example/example_output/binary_mutation_table.tsv -t example/example_output/phylogeny/phylogenetic_tree.tree -o example_output/
`````

## ML

Trains machine learning models with classification algorithms on the data and optimizes.
<br>
Available Classification algorithms: Random Forest, Support Vector Machine and Gradient Boosting

- Input, `-i`:  Binary mutation table path that is created via create_binary_tables command, can be found in create_binary_tables output path as `binary_mutation_table_with_gene_presence_absence.tsv` or `binary_mutation_table.tsv`

- Phenotype, `-p`:  Binary phenotype table path,  can be found in create_binary_tables output path as `phenotype_table.tsv` if `--create_phenotype_from_folder` is used. Can also created manually and used.

- Output, `-o`: Output folder path, where the output will be stored. If path exist, `--overwrite` option can be used to overwrite existing output.

- Antibiotic, `-a`: Antibiotic name that model will be trained. Should match the name with column that represents phenotype in binary phenotype table

- Optional arguments:
    - Machine learning algorithm, `--ml_algorithm`: Classification algorithm to be used, available selections: [rf, svm, gb]
    - Resampling strategy, `--resampling_strategy`: Resampling strategy to be used, available selections: [holdout, cv]
    - Parameter optimization, `--parameter_optimization`: Parameter optimization for model with autosklearn (https://automl.github.io/auto-sklearn/master/index.html)
    - Save model, `--save_model`: Save model
    - Feature importance analysis, `--feature_importance_analysis`: Analyze important features in the model with gini importance (for RF & GB) or permutation importance (for SVM, RF and GB)
    - Datasail, `--sail`: Splits data into training and test sets against information leakage to train better models. Requires text file that contains path of each strain per line. It can be found in create_binary_tables output path as `strains.txt` 

    More optional arguments can be found in help page: 
    `````shell
    python alpar/sr_amr.py ml -h
    `````

Basic usage:
`````shell
alpar ml -i example/example_output/binary_mutation_table.tsv -p example/example_output/phenotype_table.tsv -o example_output/ -a amikacin
`````

- `CABBAGE` data exploratory analysis lives in [EDA/cabbage.ipynb](EDA/cabbage.ipynb). 

### Windows Cursor does not show pixi / ALPAR

The pixi environment used for the ALPAR pipeline is built in **WSL (Linux)** (`.pixi/envs/default` is a Linux prefix, e.g. `x86_64-conda-linux-gnu`). Cursor running on Windows cannot use that as a kernel. That is why the kernel picker shows **Python 3.14.7** and **+ Create Python Environment**, not pixi / ALPAR.

**Python 3.14.7** in that list is usually the Windows / Microsoft Store interpreter. It will fail on `import pandas` unless you have installed packages into it. Prefer a project virtual environment.

### Use the Windows `.venv` (recommended for this notebook)

A Windows virtual environment is in `.venv/` (gitignored) with `pandas`, `pyarrow`, `matplotlib`, and `ipykernel`. It is registered as a Jupyter kernel named **Python (.venv ALPAR EDA)**.

If you need to recreate it:

```powershell
py -3 -m venv .venv
.\.venv\Scripts\python.exe -m pip install pandas pyarrow matplotlib ipykernel
.\.venv\Scripts\python.exe -m ipykernel install --user --name alpar-eda --display-name "Python (.venv ALPAR EDA)"
```

In the notebook kernel picker (top right, or **Select Kernel**):

1. Choose **Python (.venv ALPAR EDA)** if it appears.
2. If not: **Select Another Kernel…** / **Python Environments…** and pick `.venv\Scripts\python.exe`.
3. Or use **+ Create Python Environment** → **Venv** and, if Cursor offers it, the existing `.venv`.

Then run the first cell. It should print the `data\cabbage` path. Do not use bare **Python 3.14.7** unless that path is `.venv\Scripts\python.exe`.
